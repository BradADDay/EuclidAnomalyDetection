```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

# Plot formatting
cm = 0.3937008
mpl.rcParams["figure.dpi"] = 500
mpl.rcParams["figure.figsize"] = np.array((8*cm, 5*cm))
# mpl.rcParams["lines.linewidth"] = 0.2
mpl.rcParams["font.size"] = 5
mpl.rcParams["font.family"] = "serif"
mpl.rcParams["errorbar.capthick"] = 0.2
mpl.rcParams["errorbar.capsize"] = 0.5
mpl.rcParams["lines.markersize"] = 0.1

x = np.random.uniform(0, 1, 1000)
y = np.random.uniform(0, 1, 1000)
```

```python
fig, axs = plt.subplots(2, 2, sharex=True, sharey=True)

s=1

condition1 = (x-0.5)**2 + (y-0.5)**2 > 0.05
condition2 = (x-0.5)**2 + (y-0.5)**2 < 0.2

cond = condition1 & condition2

axs[0,0].scatter(x[cond], y[cond], s=s, c="k", lw=0)
axs[0,0].set_aspect("equal")
axs[0,0].set_xlim(0,1)
axs[0,0].set_ylim(0,1)
axs[0,0].set_yticks([])
axs[0,0].set_xticks([])
axs[0,0].scatter(0.55, 0.52, c="r", marker="^", s=5)
axs[0,0].vlines([0.5, 0.6], 0, 1, linestyle="--", lw=0.7, color="r")
axs[0,0].hlines([0.47, 0.57], 0, 1, linestyle="--", lw=0.7, color="r")
axs[0,0].set_ylabel("Original")

condition1 = y > 0.15*np.sin(10*x)+0.35
condition2 = y < 0.15*np.sin(10*x)+0.65

cond = condition1 & condition2

axs[0,1].scatter(x[cond], y[cond], s=s, c="k", lw=0)
axs[0,1].set_aspect("equal")
axs[0,1].set_xlim(0,1)
axs[0,1].set_ylim(0,1)
axs[0,1].set_yticks([])
axs[0,1].set_xticks([])
axs[0,1].scatter(0.55, 0.65, c="r", marker="^", s=5)
axs[0,1].vlines([0.5, 0.6], 0, 1, linestyle="--", lw=0.7, color="r")
axs[0,1].hlines([0.6], 0, 1, linestyle="--", lw=0.7, color="r")

condition1 = (x-0.5)**2 + (y-0.5)**2 > 0.05
condition2 = (x-0.5)**2 + (y-0.5)**2 < 0.2

cond = condition1 & condition2

axs[1,0].scatter(abs(x[cond]-0.5)**(1/4), abs(y[cond]-0.5)**(1/4), s=s, c="k", lw=0)
axs[1,0].set_aspect("equal")
axs[1,0].set_xlim(0,1)
axs[1,0].set_ylim(0,1)
axs[1,0].set_yticks([])
axs[1,0].set_xticks([])
axs[1,0].scatter(abs(0.55-0.5)**(1/4), abs(0.52-0.5)**(1/4), c="r", marker="^", s=5)
axs[1,0].vlines([abs(0.55-0.5)**(1/4)+0.05], 0, 1, linestyle="--", lw=0.7, color="r")
axs[1,0].hlines([abs(0.52-0.5)**(1/4)+0.05], 0, 1, linestyle="--", lw=0.7, color="r")
axs[1,0].set_ylabel("Projected")

condition1 = y > 0.15*np.sin(10*x)+0.35
condition2 = y < 0.15*np.sin(10*x)+0.65

cond = condition1 & condition2

axs[1,1].scatter(x[cond], 1.5*(y[cond]-0.15*np.sin(10*x[cond])-0.25-0.05*np.sin(3*x[cond])), s=s, c="k", lw=0)
axs[1,1].set_aspect("equal")
axs[1,1].set_xlim(0,1)
axs[1,1].set_ylim(0,1)
axs[1,1].set_yticks([])
axs[1,1].set_xticks([])
axs[1,1].scatter(0.55, 0.75, c="r", marker="^", s=5)
axs[1,1].hlines([0.7], 0, 1, linestyle="--", lw=0.7, color="r")

fig.subplots_adjust(wspace=-0.55)
fig.subplots_adjust(hspace=0.)
```
